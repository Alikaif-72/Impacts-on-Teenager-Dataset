## 8. Outlier Handing with IQR method.

```python
# Because they overcome data losses in our dataset.
```

```python
import seaborn as sns
```

```python
num_cols = ['daily_social_media_hours',
            'sleep_hours',
            'academic_performance',
            'physical_activity',
            'stress_level',
            'anxiety_level',
            'addiction_level']

for col in num_cols:

    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    df[col] = np.clip(df[col], lower, upper)
```

## 10. Features Engineering.

### (i) Mental Health Score -

```python
df['mental_health_score'] = (
    df['stress_level'] +
    df['anxiety_level'] +
    df['addiction_level']
) / 3
```

```python
# Teenager ki overall psychological condition ko represent karna.
# 3 separate mental varibles ---> i combined mental health indicator.
```

### (ii) Sleep Category -

```python
def sleep_category(x):

    if x < 5:
        return 'poor'

    elif x < 7:
        return 'average'

    else:
        return 'good'

df['sleep_category'] = df['sleep_hours'].apply(sleep_category)
```

```python
# Continuous sleep data ko human-readable groups me convert karna.
# Which sleep category has highest anxiety?
```

### (iii) Usage Category -

```python
def usage_category(x):

    if x < 2:
        return 'low'

    elif x < 5:
        return 'moderate'

    else:
        return 'high'

df['usage_category'] = df['daily_social_media_hours'].apply(usage_category)
```

```python
# Social media usage intensity identify karna.
# High-usage teenagers show higher stress levels.
```

### (iv) Wellness Score -

```python
df['wellness_score'] = (
    df['sleep_hours'] +
    df['physical_activity'] +
    df['academic_performance']
) / 3
```

```python
# Overall healthy lifestyle measure banana.
# Does social media reduce overall wellness?
```

```python
df[["mental_health_score", "wellness_score"]] = df[["mental_health_score", "wellness_score"]].round(2)
```
